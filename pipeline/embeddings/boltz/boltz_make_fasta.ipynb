```python
import pandas as pd
import os
from tqdm import tqdm

# input_path = '../../data/mhc_mapping/HLA2_IMGT_light.csv'
input_path = '../../data/unique_epitope_whole.csv'

df = pd.read_csv(input_path)
df.head()

output_path = 'input_boltz_epi'  
os.makedirs(output_path, exist_ok=True)

for i in tqdm(range(len(df))):
    # name, sequence = df['HLA_Name'][i], df['HLA_Seq'][i]
    name, sequence = df['Epi_Seq'][i], df['Epi_Seq'][i]
    # data = f">A|protein|\n{sequence}"
    data = f">A|protein|./msa/{name}.a3m\n{sequence}"
    with open(f'{output_path}/{name}.fasta', 'w') as f:
        f.write(data)
    os.makedirs(f'msa', exist_ok=True)
    with open(f'msa/{name}.a3m', 'w') as f:
        f.write(data)
```
