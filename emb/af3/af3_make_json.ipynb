```python
import pandas as pd
import os
import json
from tqdm import tqdm

def make_json(name, sequence, ptype='EPI'):
    temp = {
        "name": name,
        "sequences": [
            {
                "protein": {
                    "id": ptype,
                    "sequence": sequence
                }
            }
        ],
        "modelSeeds": [1],
        "dialect": "alphafold3",
        "version": 1
    }
    if ptype == 'EPI':
        temp["sequences"][0]["protein"].update({
            "modifications": [],
            "unpairedMsa": "",
            "pairedMsa": "",
            "templates": []
        })
    return temp

def write_json(data, name):
    with open(name, 'w') as f:
        json.dump(data, f)

def process(out_path, df, ptype, NUM_FOLDERS):
    for i in tqdm(range(len(df))):
        if ptype == 'EPI': name, sequence = df['Epi_Seq'][i], df['Epi_Seq'][i]
        elif ptype == 'HLA': name, sequence = df['HLA_Name'][i], df['HLA_Seq'][i]
        data = make_json(name, sequence, ptype)
        folder_index = i % NUM_FOLDERS  
        folder_path = f'{out_path}/{folder_index}'
        if not os.path.exists(folder_path):
            os.makedirs(folder_path, exist_ok=True)
        write_json(data, f'{folder_path}/{name}.json')
```

```python
input_path = '../../data/mhc_mapping/mhc_mapping_light.csv'
df = pd.read_csv(input_path)
out_path = 'json_hla'
ptype = 'HLA'
NUM_FOLDERS = 42
process(out_path, df, ptype, NUM_FOLDERS)
```

```python
input_path = '../../data/unique_epitope_whole.csv'
df = pd.read_csv(input_path)
out_path = 'json_epi'
ptype = 'EPI'
NUM_FOLDERS = 1
process(out_path, df, ptype, NUM_FOLDERS)
```
